```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, ConfusionMatrixDisplay
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
df = pd.read_csv("Dry_Bean.csv")
df.head()

#https://www.kaggle.com/datasets/sansuthi/dry-bean-dataset
```

```python
df["Class"].unique()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
hay_duplicados = df.duplicated().any()
hay_duplicados
```

```python
df.drop_duplicates(inplace=True)
```

```python
# revisar que el data set este balanceado 

sns.countplot(x= "Class", data=df)
plt.xticks(rotation = 45)
```

```python
# vamos a nivelar datos a bombay 

from imblearn.under_sampling import RandomUnderSampler
undersampling = RandomUnderSampler(random_state=42)
```

```python
x = df.drop("Class", axis=1)
y = df.Class
```

```python
x_over, y_over = undersampling.fit_resample(x,y)
```

```python
# revisar que el dataset este balanceado
sns.countplot(x= y_over)
plt.xticks(rotation = 45)
```

```python
df.shape
```

```python
x_over.shape
```

```python
list(np.unique(y_over))
```

```python
y_over.replace(['BARBUNYA', 'BOMBAY', 'CALI', 'DERMASON', 'HOROZ', 'SEKER', 'SIRA'], [1,2,3,4,5,6,7], inplace=True)
```

```python
list(np.unique(y_over))
```

Correlacion

```python
df_new = x_over
df_new["Class"] = y_over
```

```python
plt.figure(figsize=(15,8))
sns.heatmap(df_new.corr(), annot=True)
```

```python
x_over.drop(["ConvexArea", "EquivDiameter", "Perimeter", "MajorAxisLength", "MinorAxisLength"], axis=1, inplace=True)
```

```python
sns.pairplot(df_new, hue='Class')
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_over, y_over, test_size=0.3, random_state=42)
```

```python
model = StandardScaler()
x_train = model.fit_transform(x_train)
x_test = model.fit_transform(x_test)
```

```python
def logistic_regresion(solver_, multi_, C_parameter):
    model_regresion =  LogisticRegression(random_state=42, solver=solver_, multi_class= multi_, C= C_parameter)
    return model_regresion
```

```python
model_regresion = logistic_regresion("lbfgs", "multinomial", 1)
model_regresion.fit(x_train, y_train)
prediccion = model_regresion.predict(x_test)
print(confusion_matrix(y_test,prediccion))
print(accuracy_score(y_test,prediccion))

```

```python
solvers = ["lbfgs","newton-cg", "newton-cholesky", "sag", "saga"]
parameters = []
score   = []

for solver in solvers:
    try:
        model_regresion = logistic_regresion(solver, "multinomial", 1)
        model_regresion.fit(x_train, y_train)
        prediccion = model_regresion.predict(x_test)
        parameters.append(solver)
        accuracy = accuracy_score(y_test,prediccion)
        score.append(accuracy)
       
    except:
        None

```
