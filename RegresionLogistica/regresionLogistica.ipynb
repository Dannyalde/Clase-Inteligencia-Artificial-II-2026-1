```python
from sklearn.datasets import load_digits
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.linear_model import LogisticRegression
```

```python
digits = load_digits()
digits
```

```python
digits.data[0]
```

```python
Imagen = np.reshape(digits.data[7],(8,8)) # redimensionar la imagen a un tamaño 8x8 
```

```python
Imagen
```

```python
plt.imshow(Imagen, cmap="gray")
```

```python
digits.target[1]
```

```python
from sklearn.model_selection import train_test_split # Este es para dividir los datos en conjuntos de entrenamiento y prueba
x_train, x_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2, random_state=42)
```

```python
logistic_regresion = LogisticRegression()
logistic_regresion.fit(x_train, y_train)
```

```python
prediccion = logistic_regresion.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,prediccion)
cm
```

```python
sns.heatmap(cm, annot=True,cmap="coolwarm", linewidths=0.7)
plt.title("Matriz de confusion")
plt.ylabel("Actual label")
plt.xlabel("Predictic label")
```
