```python
import pandas as pd 

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
df.info()
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors='coerce')     
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df["PaymentMethod"]
```

```python
df.drop(columns=["customerID", "InternetService"], axis=1, inplace=True)
df.head()
```

```python
df["Churn"].replace(to_replace="No", value=0, inplace=True)
df["Churn"].replace(to_replace="Yes", value=1, inplace=True)
```

```python
df.head()
```

```python
data_procesing = df.copy()
```

```python
data_procesing = pd.get_dummies(data_procesing).astype(int)
data_procesing.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 8))
data_procesing.corr()["Churn"].sort_values(ascending = True).plot(kind="bar")
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

data_scaler = scaler.fit_transform(data_procesing)
data_scaler
```

```python
data_scaler = pd.DataFrame(data_scaler)
data_scaler.head()
```

```python
data_scaler.columns = data_procesing.columns
data_scaler.head()
```

```python
import seaborn as sns 


sns.countplot(data=df, x="gender", hue = "Churn")
```

```python
def plt_categorical_churn(column):
    sns.countplot(data=df, x=column, hue = "Churn")
    plt.show()
    
```

```python
columns = df.select_dtypes(include=object).columns
columns
```

```python
for colum in columns:
    plt_categorical_churn(colum)
```

Datos numericos 

```python
sns.pairplot(data=df, hue="Churn") # datos numericos 
```

Entrenar el modelo de regresion logistica binomial 

```python
# bonimial = hace Churn o no 
```

```python
X = data_scaler.drop("Churn", axis=1)
Y = data_scaler["Churn"].values
```

```python
from sklearn.model_selection import train_test_split 
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)


```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
prediccion = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score( y_test,prediccion)
accuracy
```

Evaluación del modelo 

```python
model.predict_proba(x_test)
```

```python
model.feature_names_in_ #categorias para saber si el modelo hace churn o no 
```

```python
pesos = pd.Series(model.coef_[0], 
                  index = X.columns.values)


plt.figure(figsize=(15,8))
pesos.sort_values(ascending =False).head(10).plot(kind="bar")
```

```python

plt.figure(figsize=(15,8))
pesos.sort_values(ascending =False)[-10:].plot(kind="bar")
```
