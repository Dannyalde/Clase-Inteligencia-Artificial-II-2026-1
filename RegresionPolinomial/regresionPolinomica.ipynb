```python
import pandas as pd 
import numpy as np
```

```python
data = pd.read_csv("Ice_cream_selling_data.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
x = data["Temperature (°C)"]
y = data["Ice Cream Sales (units)"]

X = np.array(x).reshape(-1,1)
Y = np.array(y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X,Y)
plt.title("Selling VS Temperature (°C)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.grid()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error


model = LinearRegression()
model.fit(X,Y)
prediccion_y = model.predict(X)


r2= r2_score(Y, prediccion_y)
rmse = mean_squared_error(Y, prediccion_y)**(1/2)

print(f"r2: {r2}")
print(f"rmse: {rmse}")

plt.scatter(X,Y)
plt.plot(x,prediccion_y, color = "red")
plt.title("Selling VS Temperature (°C)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.grid()
plt.show()

```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error


model2 = PolynomialFeatures(degree=17)
x_poly2= model2.fit_transform(X) 
"""Estirando el modelo para
    generar los grados del polinomio"""

model3 = LinearRegression()
model3.fit(x_poly2,Y)
prediccion_y3 = model3.predict(x_poly2)  
""""se trata como una regresion lineal de multiples variables de entrada
"""

r2= r2_score(Y, prediccion_y3)
rmse = mean_squared_error(Y, prediccion_y3)**(1/2)

print(f"r2: {r2}")
print(f"rmse: {rmse}")

plt.scatter(X,Y)
plt.plot(x,prediccion_y3, color = "red")
plt.title("Selling VS Temperature (°C)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.grid()
plt.show()

```

```python
print("Punto de corte : ", model3.intercept_ )
print("Coeficientes: ", model3.coef_)
```
